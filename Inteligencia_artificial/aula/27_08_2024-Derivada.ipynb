```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.misc import derivative

# Função para representar a paisagem
def landscape(x):
    return 0.5*(x-2)**4 + 0.7*(x-3)**3

# Função para calcular a derivada da função de paisagem
def derivative_landscape(x):
    return derivative(landscape, x, dx=1e-1)
```

```python
# Ponto de partida (ponto mais alto da paisagem)
start_x = 5.0

# Número de passos para a pessoa caminhar ao longo do gradiente
num_steps = 100

# Taxa de aprendizado para o movimento ao longo do gradiente
#learning_rate = 0.1
#learning_rate = 0.05
#learning_rate = 0.01
learning_rate = 0.001
```

```python
# Simulação da caminhada ao longo do gradiente
path_x = [start_x]

for _ in range(num_steps):
    current_x = path_x[-1]

    # Calcula a derivada no ponto atual
    gradient = derivative_landscape(current_x)
   # Atualiza o ponto com base na derivada e na taxa de aprendizado
    new_x = current_x - learning_rate * gradient

    path_x.append(new_x)
```

```python
# Plot da paisagem e da caminhada ao longo do gradiente
x = np.linspace(-2, 5, 100)
y = landscape(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y, label='Paisagem')
plt.plot(path_x, landscape(np.array(path_x)), 'ro-', label='Caminho do gradiente')
plt.xlabel('Eixo X')
plt.ylabel('Altura')
plt.title('Simulação da Paisagem')
plt.legend()
plt.grid(True)
plt.show()
```
