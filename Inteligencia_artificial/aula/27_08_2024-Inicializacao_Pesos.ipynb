```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create a simple model with a dense layer
model = Sequential([
    Dense(10, activation="relu", kernel_initializer="ones", input_shape=(2,2))
])

# Compile the model (optional for visualization)
model.compile(optimizer="adam", loss="mean_squared_error")

# Access the weights of the dense layer
dense_layer_weights = model.layers[0].get_weights()[0]
bias_values = model.layers[0].get_weights()[1]

# Print the weight values
print("Weight values:")
print(dense_layer_weights)

# Print the bias values
print("Bias values:")
print(bias_values)
```
